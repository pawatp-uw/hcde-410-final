## Does Holding Longer Reduce the Frequency of Investment Losses?

## Motivation & Problem

As someone interested in investing, I often hear that holding investments for longer reduces the risk of losing money. I want to investigate how this idea appears in historical data and how much the outcome depends on when someone starts investing. My preliminary research question is: **How does the percentage of investments ending at a loss vary across 1-month, 6-month, 1-year, 3-year, and 5-year holding periods in historical SPY ETF data?**

This question is useful because people invest for different goals and timelines. Someone saving for tuition next year may need to think differently from someone saving for a goal several years away. Average returns alone do not show how often investors might finish with less money than they started with.

I hope to compare the frequency and size of historical losses across holding periods. I could also express returns as changes in a hypothetical $1,000 investment to make the results easier to understand. The goal is to support financial understanding rather than recommend a particular investment or promise future returns.

## Data & Ethical Considerations

I plan to use the [Huge Stock Market Dataset](https://www.kaggle.com/datasets/borismarjanovic/price-volume-data-for-all-us-stocks-etfs), published by Boris Marjanovic on Kaggle. It contains historical daily stock and ETF records, including dates, opening and closing prices, daily highs and lows, and trading volume. The uploader states that prices are adjusted for stock splits and dividends.

To keep the project manageable, I will focus on the SPY ETF file, `spy.us.txt`. The inspected file contains 3,201 daily records from February 25, 2005, through November 10, 2017. It provides enough history to calculate returns over the proposed holding periods, including periods spanning the 2008 financial crisis.

The dataset is labeled **CC0: Public Domain** on Kaggle. I will document this designation and credit the source in my notebook and repository. My project code will use the MIT License required by the assignment; this does not change the dataset’s license.

The data describes market activity rather than identifiable individual investors, so personal privacy concerns are limited. However, presenting the findings responsibly is important. I will avoid implying that historical results guarantee future outcomes or that one investment timeline fits everyone. I will also distinguish losses at the end of a holding period from temporary declines experienced along the way.

## Unknowns & Dependencies

The dataset ends in 2017, so the analysis will not include more recent market conditions. I will need to check data quality further and review the documentation supporting the price adjustments.

Longer holding periods provide fewer complete observations. Overlapping periods also share market history and cannot be treated as independent experiments. I plan to compare results using a common set of starting dates as an additional check.

The initial analysis will exclude taxes, trading costs, and inflation. If data-quality issues arise, I may need to narrow the scope or find another appropriately licensed source.
