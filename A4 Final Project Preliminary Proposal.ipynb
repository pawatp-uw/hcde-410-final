## Does Holding Longer Reduce the Frequency of Investment Losses?

## Motivation & Problem

As someone interested in investing, I often hear that holding investments for longer reduces the risk of losing money. I want to investigate how this idea appears in historical data and how much the outcome depends on when someone starts investing. My preliminary research question is: **How does the percentage of investments ending at a loss vary across 1-month, 6-month, 1-year, 3-year, and 5-year holding periods in historical SPY ETF data?**

This question is useful because people invest for different goals and timelines. Someone saving for tuition next year may need to think differently from someone saving for a goal several years away. Average returns alone do not show how often investors might finish with less money than they started with.

I hope to compare the frequency and size of historical losses across holding periods. I could also express returns as changes in a hypothetical $1,000 investment to make the results easier to understand. The goal is to support financial understanding rather than recommend a particular investment or promise future returns.

# A5: Final Project Study Plan

## Overview & Research Questions / Hypotheses

This study will examine how investment holding periods relate to the frequency and size of historical losses in SPY, an ETF that tracks the S&P 500. I will use the same dataset introduced in A4, focusing on the available records from February 25, 2005, through November 10, 2017.

I am interested in whether holding an investment longer was associated with fewer losses in this historical sample. From a human-centered perspective, this matters because people invest toward goals with different deadlines. An investor who needs money for tuition soon may experience risk differently from someone saving for a goal several years away.

My primary research question is:

**How does the percentage of investments ending at a loss differ across 1-month, 6-month, 1-year, 3-year, and 5-year holding periods in this SPY dataset?**

My secondary research question is:

**How do the median return and size of the worst observed loss differ across these holding periods?**

My hypothesis is that longer holding periods will have a lower percentage of negative returns than shorter holding periods. However, I will not assume that this pattern occurs consistently or that longer holding periods eliminate risk. Findings that contradict this hypothesis will still help answer the research question.

For this study, a loss means that an investment's ending value is below its starting value, before taxes, trading costs, and inflation. This definition does not capture every type of investment risk, including temporary declines during the holding period. I will not measure these temporary declines in this study.

## Background / Related Work

The U.S. Securities and Exchange Commission's *Beginners’ Guide to Asset Allocation, Diversification, and Rebalancing* explains that investment decisions depend partly on time horizon and risk tolerance. Someone with a longer timeline may be better able to wait through market fluctuations, while someone with an upcoming expense may have less flexibility (SEC, n.d.). This helps explain why my research question is relevant to everyday investors. It also shapes how I will communicate the findings: historical outcomes should be connected to different financial timelines rather than presented as one strategy that works for everyone.

However, the relationship between time and risk is more complicated than saying that holding longer always makes stocks safer. In *On the Risk of Stocks in the Long Run*, Bodie (1995) challenges that claim by examining the cost of insuring against returns below the risk-free rate. His analysis shows that this cost can increase with the investment horizon. His measure differs from my study's measure of ending below the original investment, but it demonstrates why the definition of risk matters.

These sources motivate a focused descriptive study. I will measure historical loss frequency and the size of outcomes rather than claim to measure overall investment safety. Reporting the worst observed return alongside loss frequency will help avoid hiding the severity of unfavorable outcomes behind a single percentage.

The dataset documentation also informs the study. Marjanovic (2017) states that the historical prices are adjusted for dividends and stock splits. I will use the provided closing-price series consistently and document this reliance on the uploader's adjustment claims.

## Methodology

### Data Preparation

I will conduct the analysis in a Python Jupyter notebook using pandas for data processing and Matplotlib or Seaborn for visualizations.

I will load only the SPY file, `spy.us.txt`, and use the `Date` and `Close` columns. I will convert dates into datetime values, sort the records chronologically, and check for duplicate dates, missing values, and nonpositive prices. I will inspect unusual price changes before deciding whether any correction or exclusion is justified. Any changes will be documented, and I will preserve the original file.

### Selecting Investment Periods

I will simulate a one-time investment beginning on the first available trading day of each complete calendar month. Monthly starting dates keep the analysis manageable while allowing comparisons across different entry points.

For each starting date, I will calculate returns after 1 month, 6 months, 1 year, 3 years, and 5 years. If a target ending date falls on a weekend or market holiday, I will use the next available trading day. Periods without an available ending date will be excluded rather than estimated.

For the main comparison, I will use only starting dates with complete data for all five holding periods. This means every holding period will be evaluated using the same entry dates. Although their ending dates will differ, this approach reduces differences caused by selecting different groups of starting dates. Because the dataset ends in November 2017, the 5-year holding period limits these shared starting dates to approximately March 2005 through November 2012.


As a sensitivity check, I will repeat the analysis using all available complete periods for each holding period. I will compare the results and report the number of observations in both versions.

### Calculating and Comparing Outcomes

For each investment period, I will calculate:

**Holding-period return = (Ending adjusted closing price / Starting adjusted closing price) - 1**

A return below zero will count as a loss. A return of exactly zero will count as breaking even.

For each holding period, I will report:

- The number of complete investment periods.
- The number and percentage ending at a loss.
- The median cumulative return.
- The lowest observed return.
- The 25th and 75th percentiles of returns.

I will also translate returns into the ending value of a hypothetical $1,000 investment by multiplying $1,000 by one plus the return. This assumes no additional deposits or withdrawals.

These descriptive methods directly address the research questions without requiring a prediction model. The returns will be labeled as cumulative returns over each holding period, not annual returns.

### Limitations, Ethics, and Reproducibility

The investment periods overlap and share many of the same trading days. Therefore, they are not independent observations. I will treat the results as descriptive historical frequencies and will not use standard tests or confidence intervals that assume each period is independent.

The relatively short dataset covers one U.S. equity ETF and ends in 2017. Results may be strongly influenced by the particular market conditions included. Even if no losses appear for a holding period, that would not establish that its future loss probability is zero.

The analysis will not measure inflation-adjusted purchasing power, taxes, trading costs, or the ability of real investors to remain invested through declines. I will avoid presenting the findings as personalized financial advice.

To support reproducibility, I will document the data source, its listed CC0 designation, date-selection rules, calculations, exclusions, and software dependencies. I will include the SPY data used and the analysis notebook in the public project repository. The project code will retain the MIT License, while the dataset will retain its separate license designation.
